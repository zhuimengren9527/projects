```python
import pandas as pd


df = pd.DataFrame({
    "record_id": [101, 102, 103, 104, 105, 106],
    "site": ["R34", "R36", "R34", "R35", "R34", "R36"],
    "status": ["NORMAL", "ERROR", "ERROR", "NORMAL", "ERROR", "NORMAL"],
    "temperature": [68, 82, 91, 75, 87, 72],
    "visibility": [1200, 800, 500, 1500, 600, 1100]
})

print(df)
```

# Pandas 日练 01：选择与索引

请完成以下操作：

1. 使用 `[]` 取出 `site` 这一列，保存为 `site_series`。

```python
site_series = df['site']
site_series
```

2. 使用双层 `[]` 取出以下三列，保存为 `basic_info`：
   - `record_id`
   - `site`
   - `status`

```python
basic_info = df[['record_id','site','status']]
basic_info
```

3. 使用 `loc` 选出：
   - `site` 为 `R34`
   - 只保留 `record_id`、`status`、`temperature`
   
   保存为 `r34_records`。

```python
r34_records = df.loc[df['site'] == 'R34',['record_id','status','temperature']]
r34_records
```

4. 使用 `iloc` 取出：
   - 前 3 行
   - 第 1、2、4 列
   
   保存为 `first_three`。

```python
first_three = df.iloc[:3,[0,1,3]]
first_three
```

5. 使用 `at` 读取索引为 `2` 的记录中的 `temperature`，保存为 `temperature_at_2`。

```python
temperature_at_2 = df.at[2, "temperature"]

print(temperature_at_2)
```

6. 使用 `iat` 读取第 4 行、第 5 列的数据，保存为 `value_by_position`。

```python
value_by_position = df.iat[3, 4]

print(value_by_position)
```
