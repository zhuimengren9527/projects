````python
import pandas as pd

df = pd.DataFrame({
    "record_id": [101,102,103,104,105,106,107,108,109,110],
    "site": ["R34","R36","R34","R35","R34","R36","R35","R34","R36","R35"],
    "status": ["NORMAL","ERROR","ERROR","NORMAL","ERROR","NORMAL","ERROR","NORMAL","ERROR","ERROR"],
    "temperature": [68,82,91,75,87,72,95,69,81,88],
    "visibility": [1200,800,500,1500,600,1100,400,1300,750,550]
})

print(df)
````

### ① 使用布尔索引 `[]`

筛选：

- site 为 `"R34"`

保存为：

```python
r34_df
```

---

````python
r34_df = df[df['site'] == 'R34']
r34_df
````

### ② 使用多个条件

筛选：

- site 为 `"R34"`
- 并且 status 为 `"ERROR"`

保存为：

```python
error_df
```

````python
error_df = df[(df['site'] == 'R34') & (df['status'] == 'ERROR')]
error_df
````

### ③ 使用 `query()`

完成与②相同的筛选。

保存为：

```python
error_query
```

````python
error_query = df.query("site == 'R34' & status == 'ERROR'")
error_query
````

### ④ 使用 `isin()`

筛选：

只保留：

- R34
- R35

两个站点。

保存为：

```python
site_df
```

````python
site_df = df[df['site'].isin(['R34','R35'])]
site_df
````

### ⑤ 使用 `between()`

筛选：

temperature 在

```
70~90
```

之间（包含70和90）。

保存为：

```python
temp_df
```

````python
temp_df = df[df['temperature'].between(70,90)]
temp_df
````

### ⑥ 使用 `where()`

将：

temperature 小于 80

的数据替换为 NaN。

保存为：

```python
temp_where
```

````python
temp_where = df.where(df['temperature'] >= 80)
temp_where
````

### ⑦ 使用 `mask()`

将：

temperature 大于等于 80

的数据替换为 NaN。

保存为：

```python
temp_mask
```

````python
temp_mask = df.mask(df['temperature'] >= 80)
temp_mask
````
