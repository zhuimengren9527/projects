````python
import pandas as pd

df = pd.DataFrame({
    "record_id":[101,102,103,104,105,106,107,108,109,110,111,112],
    "site":["R34","R36","R35","R34","R35","R36","R34","R35","R36","R34","R35","R34"],
    "status":["ERROR","NORMAL","ERROR","ERROR","NORMAL","ERROR","ERROR","ERROR","NORMAL","ERROR","ERROR","NORMAL"],
    "temperature":[72,65,88,91,75,83,79,86,73,70,92,81],
    "visibility":[800,1200,550,680,900,450,520,610,1000,580,700,1300]
})

print(df)
````

### 任务1

只分析：

- R34
- R35

两个站点。

````python
df = df[df['site'].isin(['R34','R35'])]
df
````

### 任务2

只保留：

状态为

```
ERROR
```

的数据。

---

````python
df = df[df['status'] == 'ERROR']
df
````

### 任务3

仪器规定：

只有温度在

```
70~90℃
```

````python
df = df[df['temperature'].between(70,90)]
df
````

### 任务4

由于低能见度容易产生误报警。

因此：

对于 `visibility < 600` 的记录，不要删除，

而是将

```
visibility
```

修改为缺失值（NaN）。

---

````python
df['visibility'] =df['visibility'].mask(df['visibility'] < 600)
df
````
