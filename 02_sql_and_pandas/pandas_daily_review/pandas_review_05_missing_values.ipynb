````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "record_id":[101,102,103,104,105,106,107,108,109,110],
    "site":["R34","R36","R35","R34","R35","R36","R34","R35","R36","R34"],
    "status":["ERROR","NORMAL",np.nan,"ERROR","NORMAL","ERROR","ERROR",np.nan,"NORMAL","ERROR"],
    "temperature":[72,np.nan,88,91,np.nan,83,79,86,73,np.nan],
    "visibility":[800,1200,np.nan,680,900,450,np.nan,610,1000,580]
})

print(df)
````

### 任务1

统计每一列缺失值的数量。

保存为：

```python
missing_count
```

````python
missing_count = df.isna().sum()
missing_count
````

### 任务2

找出：

**temperature 缺失**

或者

**visibility 缺失**

的所有记录。

保存为：

```python
missing_records
```

---

````python
missing_records = df[
    df["temperature"].isna()
    | df["visibility"].isna()
]

missing_records
````

### 任务3

对于：

temperature 缺失

使用

```
同站点(site)的平均温度
```

进行填充。

保存结果为：

```python
filled_df
```

---

````python
filled_df = (
    df
    .assign(
        avg_temp=lambda x: (
            x.groupby("site")["temperature"]
             .transform("mean")
        ),
        temperature=lambda x: (
            x["temperature"].fillna(x["avg_temp"])
        )
    )
    .drop(columns="avg_temp")
)
filled_df
````

### 任务4

由于：

status

属于业务字段。

如果：

status 缺失

认为这条数据无法分析。

请删除这些记录。

保存为：

```python
clean_df
```

---

````python
clean_df = (
    filled_df
    .dropna(subset=["status"])
    .reset_index(drop=True)
)

clean_df
````
