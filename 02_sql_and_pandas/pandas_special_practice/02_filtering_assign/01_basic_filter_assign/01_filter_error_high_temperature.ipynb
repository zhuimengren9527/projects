## 练习 05：标记需要重点检查的设备

### 任务

创建 `df_raw` 的副本：

```python
df_result
```

在 `df_result` 中新增字段：

```text
inspection_flag
```

先将所有记录的 `inspection_flag` 设置为：

```text
NORMAL
```

然后将同时满足以下条件的记录修改为：

```text
CHECK
```

条件：

- `status` 等于 `ERROR`
- `temperature` 大于或等于 80

### 要求

- 不允许直接修改 `df_raw`。
- 必须使用 `.copy()` 创建 `df_result`。
- 必须使用布尔掩码表示条件。
- 两个条件之间使用 `&`。
- 每个条件分别放在括号中。
- 必须使用 `.loc[]` 完成条件赋值。
- 不使用循环。
- 不使用 `apply()`。
- 最后输出完整的 `df_result`。

````python
df_result = df_raw.copy()

df_result['inspection_flag'] = 'NORMAL'

check_mask = (df_result['status'] == 'ERROR') & (df_result['temperature'] >= 80)

df_result.loc[check_mask,'inspection_flag'] = 'CHECK'

df_result


````
