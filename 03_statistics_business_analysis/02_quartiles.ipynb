# 统计学基础 02：四分位数

## 任务

下面是一组设备故障持续时间，单位为分钟：

```text
5、7、8、9、10、11、12、13、15、18、60
```

请使用 Pandas 计算：

- 第一四分位数 `Q1`，即 25% 分位数
- 第二四分位数 `Q2`，即 50% 分位数
- 第三四分位数 `Q3`，即 75% 分位数

将结果分别保存为：

```python
q1
q2
q3
```

最后用一句话说明：

> 这组故障中，大约 75% 的故障持续时间不超过多少分钟？

## 要求

- 使用 `.quantile()`。
- 不手工计算。
- 结果保留两位小数。

````python
import pandas as pd

duration = pd.Series([
    5, 7, 8, 9, 10, 11, 12, 13, 15, 18, 60
])

duration
````

````python
q1 = duration.quantile(0.25)
q2 = duration.quantile(0.50)
q3 = duration.quantile(0.75)

print(f"Q1：{q1:.2f}")
print(f"Q2：{q2:.2f}")
print(f"Q3：{q3:.2f}")
````
