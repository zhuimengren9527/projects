```python
# 读取并存储数据
from pathlib import Path
df = pd.read_csv('../day1_missing_values/data/raw_data/Building_Permits.csv')
print(df.head())

# 存储df到今日练习文件夹
data_dir = Path.cwd()/'data'/'raw_data'
data_dir.mkdir(parents=True,exist_ok=True)
file_path = data_dir/'Building_Permits.csv'
df.to_csv(file_path,index=False)
```

```python
# 准备数据
original_data = df['Estimated Cost']
print(original_data.head())

# 看该列数据是否有空值
print(original_data.isna().sum())

# 删除空值
original_data = original_data.dropna()
print(original_data.isna().sum())
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from scipy import stats # 用于真正的 Normalization (Box-Cox)


# 1. MinMaxScaler (缩放到 0-1)
scaler_minmax = MinMaxScaler()
data_minmax = scaler_minmax.fit_transform(original_data)

# 2. StandardScaler (标准化到均值0, 标准差1)
scaler_std = StandardScaler()
data_std = scaler_std.fit_transform(original_data)

# 3. Normalization (Box-Cox 变换，这才是 Day 2 追求的“正态化”)
# 注意：Box-Cox 要求数据必须大于 0
data_normalized = stats.boxcox(original_data['Estimated Cost'])[0]
```
