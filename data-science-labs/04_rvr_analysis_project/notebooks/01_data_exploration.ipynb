```python
import pandas as pd
# 调整显示设置
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
pd.set_option('display.max_colwidth', 25)
df = pd.read_csv('../data/raw/VIS_R34_26.his',sep='\s+',skiprows=1,encoding='latin1',on_bad_lines='skip')

print(df.shape)
print(df.describe())
print(df.head(-6))
```

```python
# 1.确保时间列是时间格式
df['CREATEDATE'] = pd.to_datetime(df['CREATEDATE'])

# 2.计算时间差(diff)
df['time_diff'] = df['CREATEDATE'].diff().dt.total_seconds()

# 3.找出所有异常间隙
missing_data = df[df['time_diff']>15]

# 4.打印结果
print("发现以下疑似掉包的时间节点：")
print(missing_data[['CREATEDATE','time_diff']])
```
