```python
# 设置数据显示格式
import pandas as pd
pd.set_option('display.expand_frame_repr',False)
pd.set_option('display.max_columns',None)
```

```python
# 读取数据
df = pd.read_csv('../day1_missing_values/data/raw_data/Building_Permits.csv')
df.head()
```

```python
print(df.shape)
print(f"数据共有{df.shape[0]}行，{df.shape[1]}列")
```

```python
# 计算每列的缺失值数量
missing_values_count = df.isnull().sum()
print(missing_values_count)

# 计算整个表的总缺失值
total_missing_values = missing_values_count.sum()
print(total_missing_values)

# 计算数据的总格数
import numpy as np
total_cells = np.prod(df.shape)
print(total_cells)

# 计算缺失值再总数据中的占比
percent_missing = (total_missing_values/total_cells)*100
print(f"缺失值占比为:{percent_missing : .1f}%")
```

```python
# 尝试删除所有包含缺失值的行
df_cleaned_rows = df.dropna()
print(f"原始行数:{df.shape[0]}")
print(f"删除后剩下的行数:{df_cleaned_rows.shape[0]}")
```

```python
# 尝试删除所有包含缺失值的列
df_cleaned_col = df.dropna(axis=1)
print(f"原始列数:{df.shape[1]}")
print(f"删除后剩下的列数:{df_cleaned_col.shape[1]}")
```

```python
# 把所有空值替换成0
df_null_with_zero = df.fillna(0)

# 计算数据空值
df_filled_zero = df_null_with_zero.isna().sum()
print(df_filled_zero.sum())
```

```python
# 用后一个值填补空值
df_null_with_bfill = df.bfill(axis=0).fillna(0)

# 计算数据空值
df_filled_bfill = df_null_with_bfill.isna().sum()
print(df_filled_bfill.sum())
```
