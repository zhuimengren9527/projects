```python
import sqlite3
import pandas as pd

# 统一锁死的绝对路径
DB_PATH = r"C:\projects\machinelearning_study\raw_data\datacamp\SQL\world.db"

try:
    conn = sqlite3.connect(DB_PATH)
    
    # 设置 Pandas 打印参数，确保多列数据在终端里完美对齐不折行
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    pd.set_option('display.colheader_justify', 'center')
    
    print("=" * 60)
    print(f" 📂 成功连接数据库，正在打印全新沙盒中的三张表... ")
    print("=" * 60)
    
    # 1. 打印国家表
    print("\n📊 【1. 国家表: countries】")
    print("-" * 50)
    countries_df = pd.read_sql_query("SELECT * FROM countries;", conn)
    print(countries_df.to_string(index=False))
    
    # 2. 打印城市表
    print("\n📊 【2. 城市表: cities】")
    print("-" * 50)
    cities_df = pd.read_sql_query("SELECT * FROM cities;", conn)
    print(cities_df.to_string(index=False))
    
    # 3. 打印语言表
    print("\n📊 【3. 语言表: languages】")
    print("-" * 50)
    languages_df = pd.read_sql_query("SELECT * FROM languages;", conn)
    print(languages_df.to_string(index=False))
    
    print("\n" + "=" * 60)
    print(" 地基已彻底对齐！请确认打印无误后，随时带上你的‘环境通过’信号！")
    print("=" * 60)

except Exception as e:
    print(f"\n❌ 打印失败，请检查数据创建脚本是否成功运行。错误详情: {e}")

finally:
    if 'conn' in locals():
        conn.close()
```
