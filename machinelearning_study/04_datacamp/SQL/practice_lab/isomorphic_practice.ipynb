```python
import sqlite3
import pandas as pd

data_path = r"C:\projects\machinelearning_study\raw_data\datacamp\SQL\world.db"
conn = sqlite3.connect(data_path)

```

```python
# SQL轨道
sql_query = """
SELECT co.code AS country_code,
       co.name AS country_name,
       COALESCE(SUM(ci.population),0)  AS total_urban_population,
      CASE WHEN COALESCE(SUM(ci.population), 0) < 1000000 THEN 1 ELSE 0 END AS is_high_risk_profile
FROM countries AS co
LEFT JOIN cities AS ci
ON co.code = ci.country_code
GROUP BY co.code, co.name
"""
sql_result = pd.read_sql_query(sql_query,conn)
print(sql_result)
```

```python
# pandas轨道
df_cities = pd.read_sql_query("SELECT * FROM cities;",conn)
df_counties = pd.read_sql_query("SELECT * FROM countries;",conn)

df_pandas_merged = pd.merge(
    df_counties,
    df_cities,
    left_on='code',
    right_on='country_code',
    how = 'left',
    suffixes=('_country','_city')
).drop('country_code',axis=1)
df_pandas_merged_clean = df_pandas_merged.fillna({'population':0})
print(df_pandas_merged_clean)
```

```python
assert len(sql_result) == len(df_pandas_merged_clean)
print(f"✅验证通过")
```
