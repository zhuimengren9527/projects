```python
import pandas as pd

# 准备数据
orders = pd.DataFrame({
    'order_id': [1, 2, 3, 4],
    'p_id': [101, 102, 101, 105], # 注意：105 在产品表中不存在
    'qty': [2, 5, 1, 10]
})

products = pd.DataFrame({
    'p_id': [101, 102, 103, 104],
    'name': ['鼠标', '键盘', '显卡', '显示器'],
    'price': [50, 150, 3000, 1200]
})
print(f"打印表‘orders’:\n{orders}")
print(f"打印表‘products’:\n{products}")
```

```python
# 深入第一步：Inner Join (默认)
# 逻辑：取交集。注意 order_id 4 消失了，因为 p_id 105 没在 products 里。
inner_results = pd.merge(orders, products, on='p_id', how='inner',validate='many_to_one')
print(inner_results)
```

```python
audit_res = pd.merge(orders,products,on='p_id',how='left',indicator=True,validate='many_to_one')
print(audit_res)
```
