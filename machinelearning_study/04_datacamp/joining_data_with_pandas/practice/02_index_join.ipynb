```python
import pandas as pd

# 准备已经设好索引的数据
orders_idx = pd.DataFrame({
    'order_id': [1, 2, 3]
}, index=[101, 102, 101]) # p_id 作为索引
orders_idx.index.name = 'p_id'

products_idx = pd.DataFrame({
    'price': [50.0, 150.0]
}, index=[101, 102]) # p_id 作为索引
products_idx.index.name = 'p_id'

# 极其简洁的 join 语法
res = orders_idx.join(products_idx)

print(res)
```
